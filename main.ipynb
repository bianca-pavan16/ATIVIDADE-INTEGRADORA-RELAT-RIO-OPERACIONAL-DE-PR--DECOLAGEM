```python
#Os parâmetros utilizados foram definidos previamente
#para a simulação:
#Temperatura interna: 15 °C a 30 °C #
#Temperatura externa: 0 °C a 40 °C #
#Integridade estrutural: deve ser igual a 1 #
#Nível de energia: mínimo de 80% #
#Pressão dos tanques: 200 a 300 PSI #
#Módulos críticos: todos devem ser iguais a 1


# criação de variáveis usando o input para ser inserido o valor"

Temp_Int= float(input("Qual o valor em Celcius da TEMPERATURA INTERNA da nave?"))
Temp_Ext=float(input("Qual o valor em Celcius da TEMPERATURA EXTERNA da nave?"))
Integ_Estrutural=int(input("Sendo 1 operante, e 0 como inoperante, como está a INTEGRIDADE ESTRUTURAL do foguete?"))
Nivel_Energia=float(input("em quantos % está os NÍVEIS DE ENERGIA para decolagem?"))
Pressao_Tanque=float(input("Qual o valor em PSI da PRESSÃO DOS TANQUES??"))
print("\nSTATUS DOS MÓDULOS CRÍTICOS")
Propulsao=int(input("Sendo 1 operante, e 0 como inoperante, como está a PROPULSÃO do foguete?"))
Comunicacao=int(input("Sendo 1 operante, e 0 como inoperante, como está a COMUNICAÇÃO do foguete?"))
Navegacao=int(input("Sendo 1 operante, e 0 como inoperante, como está a NAVEGAÇÃO do foguete?"))



#criação das condições usando IF
# Variável de controle - começa assumindo que está tudo certo
pronto = True

print("\n--- VERIFICAÇÃO DOS SISTEMAS ---")

#temperatura interna
if Temp_Int < 15 or Temp_Int > 30:
    print("- Temperatura interna fora da faixa segura (15°C a 30°C)")
    pronto = False

#temperatura externa
if Temp_Ext < 0 or Temp_Ext > 40:
    print("- Temperatura externa fora da faixa segura (0°C a 40°C)")
    pronto = False

#Integridade Estrutural
if Integ_Estrutural != 1:
    print("- Integridade estrutural comprometida")
    pronto = False

#Nível de energia
if Nivel_Energia < 80:
    print("- Nível de energia insuficiente (mínimo 80%)")
    pronto = False

#Pressão tanque
if Pressao_Tanque < 200 or Pressao_Tanque > 300:
    print("- Pressão dos tanques fora da faixa segura (200 a 300 PSI)")
    pronto = False

#Modulos Críticos

if Propulsao != 1:
    print("- Módulo de propulsão inoperante")
    pronto = False

if Comunicacao != 1:
    print("- Módulo de comunicação inoperante")
    pronto = False

if Navegacao != 1:
    print("- Módulo de navegação inoperante")
    pronto = False

# Decisão final
print("\n--- RESULTADO FINAL ---")
if pronto == True:
    print(
    f"Com a temperatura interna igual a {Temp_Int}°C  "
    f"\n a temperatura externa igual a {Temp_Ext}°C, "
    f"\n a integridade estrutural estando OK."
    f"\n O nível de energia igual a {Nivel_Energia}%, "
    f"\n a pressão dos tanques {Pressao_Tanque} PSI."
    f"\n Os módulos críticos de comunicação, navegação "
    f"\n e propulsão estão OK."
)
    print("\nSTATUS: PRONTO PARA DECOLAR")
else:
    print("STATUS: DECOLAGEM ABORTADA")


    #Criação de cenários para teste
cenarios = [
    {
        "nome": "Cenário 1 - Condição normal",
        "Temp_Int": 22,
        "Temp_Ext": 22,
        "Integ_Estrutural": 1,
        "Nivel_Energia": 88,
        "Pressao_Tanque": 222,
        "Propulsao": 1,
        "Comunicacao": 1,
        "Navegacao": 1
    },

    {
        "nome": "Cenário 2 - Temperatura interna elevada",
        "Temp_Int": 35,
        "Temp_Ext": 22,
        "Integ_Estrutural": 1,
        "Nivel_Energia": 88,
        "Pressao_Tanque": 222,
        "Propulsao": 1,
        "Comunicacao": 1,
        "Navegacao": 1
    },

    {
        "nome": "Cenário 3 - Temperatura externa elevada",
        "Temp_Int": 22,
        "Temp_Ext": 45,
        "Integ_Estrutural": 1,
        "Nivel_Energia": 88,
        "Pressao_Tanque": 222,
        "Propulsao": 1,
        "Comunicacao": 1,
        "Navegacao": 1
    },

    {
        "nome": "Cenário 4 - Energia insuficiente",
        "Temp_Int": 22,
        "Temp_Ext": 22,
        "Integ_Estrutural": 1,
        "Nivel_Energia": 65,
        "Pressao_Tanque": 222,
        "Propulsao": 1,
        "Comunicacao": 1,
        "Navegacao": 1
    },

    {
        "nome": "Cenário 5 - Pressão dos tanques baixa",
        "Temp_Int": 22,
        "Temp_Ext": 22,
        "Integ_Estrutural": 1,
        "Nivel_Energia": 88,
        "Pressao_Tanque": 150,
        "Propulsao": 1,
        "Comunicacao": 1,
        "Navegacao": 1
    },

    {
        "nome": "Cenário 6 - Pressão dos tanques elevada",
        "Temp_Int": 22,
        "Temp_Ext": 22,
        "Integ_Estrutural": 1,
        "Nivel_Energia": 88,
        "Pressao_Tanque": 350,
        "Propulsao": 1,
        "Comunicacao": 1,
        "Navegacao": 1
    },

    {
        "nome": "Cenário 7 - Falha de integridade estrutural",
        "Temp_Int": 22,
        "Temp_Ext": 22,
        "Integ_Estrutural": 0,
        "Nivel_Energia": 88,
        "Pressao_Tanque": 222,
        "Propulsao": 1,
        "Comunicacao": 1,
        "Navegacao": 1
    },

    {
        "nome": "Cenário 8 - Falha no módulo de propulsão",
        "Temp_Int": 22,
        "Temp_Ext": 22,
        "Integ_Estrutural": 1,
        "Nivel_Energia": 88,
        "Pressao_Tanque": 222,
        "Propulsao": 0,
        "Comunicacao": 1,
        "Navegacao": 1
    },

    {
        "nome": "Cenário 9 - Falha no módulo de comunicação",
        "Temp_Int": 22,
        "Temp_Ext": 22,
        "Integ_Estrutural": 1,
        "Nivel_Energia": 88,
        "Pressao_Tanque": 222,
        "Propulsao": 1,
        "Comunicacao": 0,
        "Navegacao": 1
    },

    {
        "nome": "Cenário 10 - Múltiplas anomalias",
        "Temp_Int": 38,
        "Temp_Ext": 22,
        "Integ_Estrutural": 1,
        "Nivel_Energia": 60,
        "Pressao_Tanque": 340,
        "Propulsao": 1,
        "Comunicacao": 0,
        "Navegacao": 1
    }
]


for cenario in cenarios:

    pronto = True
    falhas = []

    if cenario["Temp_Int"] < 15 or cenario["Temp_Int"] > 30:
        pronto = False
        falhas.append("Temperatura interna fora da faixa segura")

    if cenario["Temp_Ext"] < 0 or cenario["Temp_Ext"] > 40:
        pronto = False
        falhas.append("Temperatura externa fora da faixa segura")

    if cenario["Integ_Estrutural"] != 1:
        pronto = False
        falhas.append("Falha na integridade estrutural")

    if cenario["Nivel_Energia"] < 80:
        pronto = False
        falhas.append("Nível de energia insuficiente")

    if cenario["Pressao_Tanque"] < 200 or cenario["Pressao_Tanque"] > 300:
        pronto = False
        falhas.append("Pressão dos tanques fora da faixa segura")

    if cenario["Propulsao"] != 1:
        pronto = False
        falhas.append("Falha no módulo de propulsão")

    if cenario["Comunicacao"] != 1:
        pronto = False
        falhas.append("Falha no módulo de comunicação")

    if cenario["Navegacao"] != 1:
        pronto = False
        falhas.append("Falha no módulo de navegação")

    print("\n----------------------------------------")
    print(cenario["nome"])

    if pronto:
        print("STATUS: PRONTO PARA DECOLAR")
    else:
        print("STATUS: DECOLAGEM ABORTADA")
        print("Anomalias detectadas:")

        for falha in falhas:
            print("-", falha)

# 1.4 - Análise Energética

# Valores adotados para a simulação
capacidade_total = 500       # kWh
consumo_decolagem = 100      # kWh
perda_percentual = 10        # %
consumo_medio = 50           # kW
carga_minima = 80            # %


print("=== ANÁLISE ENERGÉTICA ===")
print("\nEscolha uma situação para analisar:")

# Mostra os cenários que já foram criados anteriormente
for i in range(len(cenarios)):
    print(i + 1, "-", cenarios[i]["nome"])


# Usuário escolhe o cenário
opcao = int(input("\nDigite o número da situação: "))

# Pega o cenário escolhido
cenario = cenarios[opcao - 1]

# Pega o nível de energia do cenário
carga_atual = cenario["Nivel_Energia"]


# -----------------------------
# CÁLCULOS ENERGÉTICOS
# -----------------------------

# Energia disponível de acordo com a carga atual
energia_disponivel = capacidade_total * (carga_atual / 100)

# Cálculo das perdas energéticas
perda_energia = energia_disponivel * (perda_percentual / 100)

# Energia realmente disponível após as perdas
energia_util = energia_disponivel - perda_energia

# Autonomia inicial
autonomia_inicial = energia_util / consumo_medio

# Energia restante depois da decolagem
energia_restante = energia_util - consumo_decolagem

# Autonomia após a decolagem
if energia_restante > 0:
    autonomia_apos_decolagem = energia_restante / consumo_medio
else:
    autonomia_apos_decolagem = 0


# -----------------------------
# RESULTADOS
# -----------------------------

print("\n===================================")
print("Situação:", cenario["nome"])
print("===================================")

print("Capacidade total:", capacidade_total, "kWh")
print("Carga atual:", carga_atual, "%")

print("\n--- Análise de energia ---")
print("Energia disponível:", round(energia_disponivel, 2), "kWh")
print("Perdas energéticas:", round(perda_energia, 2), "kWh")
print("Energia útil:", round(energia_util, 2), "kWh")

print("\n--- Autonomia ---")
print("Consumo médio:", consumo_medio, "kW")
print("Autonomia inicial:", round(autonomia_inicial, 2), "horas")

print("\n--- Decolagem ---")
print("Consumo estimado na decolagem:", consumo_decolagem, "kWh")
print("Energia restante:", round(energia_restante, 2), "kWh")
print("Autonomia após decolagem:",
      round(autonomia_apos_decolagem, 2), "horas")


# -----------------------------
# VERIFICAÇÃO PARA DECOLAGEM
# -----------------------------

if carga_atual < carga_minima:
    print("\nSTATUS: DECOLAGEM ABORTADA")
    print("Motivo: carga abaixo do mínimo de 80%")

elif energia_restante <= 0:
    print("\nSTATUS: DECOLAGEM ABORTADA")
    print("Motivo: energia insuficiente para realizar a decolagem")

else:
    print("\nSTATUS: ENERGIA ADEQUADA PARA DECOLAGEM")
```
